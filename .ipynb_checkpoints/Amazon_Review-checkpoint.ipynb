# Executive Summary: Amazon Customer Reviews Analysis
As a data scientist for an analytics firm, I am tasked with extracting actionable business intelligence from a dataset of 997 customer reviews. This project demonstrates a cohesive data science workflow across three distinct domains:
1. **Natural Language Processing (NLP):** Identifying core product features and customer pain points from unstructured text.
2. **Time Series Analysis:** Tracking review volume and customer satisfaction trends over a 14-year period.
3. **Neural Networks:** Building a predictive model to automatically classify the sentiment (star rating) of future reviews based solely on text content.

**Initial Observation:** The dataset suffers from a severe class imbalance, with 580 reviews (58%) rated as 5-stars, and only 48 reviews rated as 2-stars. This skew will significantly impact predictive modeling later in the pipeline.

## Part 1: NLP

In this section, I will clean and preprocess review text, explore frequent words and bigrams, and visualize key language patterns in the dataset.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from collections import Counter
from nltk.util import ngrams
import warnings

# Suppress generic warnings
warnings.filterwarnings('ignore')

# Download required NLTK resources silently
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)

# Load the dataset
df = pd.read_csv('amazon_reviews_lab.csv')

# Explore the dataset
print(f"Dataset Shape: {df.shape}")
print("\nRating Value Counts:\n", df['rating'].value_counts())
print("\nFirst 3 rows:\n", df.head(3))

# Create preprocessing tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    # 1. Convert to lowercase
    text = str(text).lower()
    # 2. Remove URLs
    text = re.sub(r'http\S+|www\.\S+', '', text)
    # 3. Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)
    # 4. Tokenize
    tokens = word_tokenize(text)
    # 5. Keep alphabetic tokens only
    tokens = [word for word in tokens if word.isalpha()]
    # 6. Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]
    # 7. Lemmatize
    cleaned_tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return cleaned_tokens

# Apply preprocessing
df['processed_tokens'] = df['review_text'].apply(preprocess_text)

# Combine all tokens into one list
all_tokens = [token for token_list in df['processed_tokens'] for token in token_list]

# Identify most common words and bigrams
word_counts = Counter(all_tokens)
print("\nTop 10 Words:", word_counts.most_common(10))

all_bigrams = list(ngrams(all_tokens, 2))
bigram_counts = Counter(all_bigrams)
print("Top 10 Bigrams:", bigram_counts.most_common(10))

# Visualize top 20 words
top_20_words = word_counts.most_common(20)
words, counts = zip(*top_20_words)

plt.figure(figsize=(12, 6))
sns.barplot(x=list(counts), y=list(words), palette="viridis")
plt.title("Top 20 Most Frequent Words in Reviews")
plt.xlabel("Frequency")
plt.ylabel("Words")
plt.show()
```

### NLP Insights: What are customers talking about?
By passing the raw reviews through a preprocessing pipeline (tokenization, stopword removal, and lemmatization), we isolated the semantic core of the customer feedback. 

* **Feature Extraction:** The frequency analysis tells a clear story about the product category. Top terms like 'nook', 'kindle', 'book', and 'screen' dominate the discourse. 
* **Customer Priorities:** The bigram analysis is particularly insightful for product teams. Phrases like 'battery life', 'touch screen', 'sd card', and 'work great' highlight the specific hardware features that drive customer satisfaction and purchasing decisions.

## Part 2: Time Series

In this section, I will analyze review activity over time by converting timestamps, resampling the data, and visualizing trends in review volume and ratings.


```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset (if running independently of Part 1)
df = pd.read_csv('amazon_reviews_lab.csv')

# Convert timestamp to datetime format
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Inspect date range
print(f"Earliest review: {df['timestamp'].min()}")
print(f"Latest review: {df['timestamp'].max()}")

# Set timestamp as index for time series operations
df.set_index('timestamp', inplace=True)

# Resample by month
monthly_review_count = df.resample('M').size()
monthly_avg_rating = df['rating'].resample('M').mean()

# Plot 1: Monthly Review Volume
plt.figure(figsize=(12, 5))
monthly_review_count.plot(color='blue')
plt.title('Monthly Review Volume Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Reviews')
plt.grid(True)
plt.show()

# Plot 2: Average Monthly Rating with 3-Month Rolling Average
rolling_avg_rating = monthly_avg_rating.rolling(window=3).mean()

plt.figure(figsize=(12, 5))
monthly_avg_rating.plot(label='Original Monthly Avg', color='lightgray', marker='o')
rolling_avg_rating.plot(label='3-Month Rolling Avg', color='red', linewidth=2)
plt.title('Average Monthly Rating Over Time')
plt.xlabel('Date')
plt.ylabel('Average Rating (1-5)')
plt.legend()
plt.grid(True)
plt.show()
```

### Temporal Trends: Review Volume vs. Satisfaction
Visualizing the data chronologically from May 2000 to July 2014 reveals how customer engagement evolves over a product's lifecycle.

* **Volume Volatility:** The first chart demonstrates that review volume is not consistent; it features sharp spikes, which typically correlate with holiday shopping seasons, major sales events (like Prime Day), or new product launches. 
* **Smoothing the Noise:** Daily or monthly average ratings can be highly volatile due to low sample sizes in certain periods. By applying a **3-Month Rolling Average** (the red line), we successfully smoothed out short-term noise to reveal the true underlying trend in customer satisfaction. If the rolling average drops during a volume spike, it signals that the supply chain or customer support may be struggling to handle peak demand.

```python
from statsmodels.tsa.stattools import adfuller

# 1. Formal Stationarity Test (Augmented Dickey-Fuller)
print("--- Original Monthly Review Volume ---")
adf_result = adfuller(monthly_review_count)
print(f"ADF Statistic: {adf_result[0]:.4f}")
print(f"p-value: {adf_result[1]:.4f}")

# 2. Data Transformation (First-Order Differencing)
monthly_diff = monthly_review_count.diff().dropna()

print("\n--- Differenced Monthly Review Volume ---")
adf_diff_result = adfuller(monthly_diff)
print(f"ADF Statistic: {adf_diff_result[0]:.4f}")
print(f"p-value: {adf_diff_result[1]:.4f}")

# Visualize the stationary series
plt.figure(figsize=(12, 4))
monthly_diff.plot(color='green')
plt.title('Stationary Time Series: First-Order Differenced Monthly Reviews')
plt.axhline(0, color='black', linestyle='--', linewidth=1)
plt.ylabel('Differenced Count')
plt.show()
```

### Stationarity & Transformation Analysis
To build a reliable forecasting model, the time series must be stationary (constant mean and variance). 
* **Initial Test:** The Augmented Dickey-Fuller (ADF) test on the original monthly review volume returned a p-value greater than 0.05, confirming the data is non-stationary due to trend volatility.
* **Transformation:** By applying first-order differencing, we stabilized the mean. The subsequent ADF test on the differenced data returned a p-value below the 0.05 threshold, formally rejecting the null hypothesis and confirming the series is now stationary and ready for autoregressive modeling.

```python
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

# 1. Train/Test Split (80/20 chronological split)
train_size = int(len(monthly_review_count) * 0.8)
train, test = monthly_review_count.iloc[:train_size], monthly_review_count.iloc[train_size:]

# 2. Model Selection & Fitting: ARIMA(1,1,1)
# d=1 accounts for the differencing applied earlier
model = ARIMA(train, order=(1, 1, 1))
fitted_model = model.fit()

# Display model diagnostics (AIC/BIC)
print(fitted_model.summary().tables[0])

# 3. Forecasting
predictions = fitted_model.forecast(steps=len(test))

# 4. Error Metrics
rmse = np.sqrt(mean_squared_error(test, predictions))
mae = mean_absolute_error(test, predictions)
print(f'\nForecast Evaluation -> RMSE: {rmse:.2f} | MAE: {mae:.2f}')

# 5. Visual Evaluation
plt.figure(figsize=(12, 5))
train.plot(label='Training Data')
test.plot(label='Actual Test Data', color='orange')
predictions.plot(label='ARIMA Forecast', color='red', linestyle='--')
plt.title('ARIMA Forecasting: Expected vs. Actual Review Volume')
plt.legend()
plt.show()
```

### Forecast Evaluation & Business Implications
We implemented an ARIMA(1,1,1) model, selected because the data required one order of differencing (d=1) to achieve stationarity, paired with baseline autoregressive (p=1) and moving average (q=1) terms to capture recent momentum and shock residuals.

* **Diagnostic Metrics:** The model's AIC and BIC scores confirm a parsimonious fit without over-parameterization.
* **Performance:** The Root Mean Squared Error (RMSE) and Mean Absolute Error (MAE) quantify the average deviation between our forecasted review volume and the actual test data. 
* **Business Implication:** The forecast successfully captures the baseline trajectory of customer engagement but struggles to predict sudden, extreme spikes (such as random viral events or unannounced sales). For the analytics client, this baseline forecast is highly actionable: it allows the operations team to appropriately scale customer support staffing for expected baseline volume, while highlighting that anomaly detection systems are needed to handle sudden volume shocks.

## Part 3: Neural Networks

In this section, I will transform text reviews into numerical features, train a neural network to predict ratings, and evaluate model performance.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.utils.class_weight import compute_class_weight
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers, callbacks

# Load dataset
df = pd.read_csv('amazon_reviews_lab.csv')
df = df.dropna(subset=['review_text', 'rating'])

# Define X and y (shift ratings 1-5 → 0-4)
X = df['review_text']
y = df['rating'] - 1

# TF-IDF vectorization
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
X_tfidf = vectorizer.fit_transform(X)

# Train / test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=42, stratify=y
)

# Class weights (imbalance handling)
classes = np.unique(y_train)
class_weights = compute_class_weight('balanced', classes=classes, y=y_train)
class_weight_dict = dict(zip(classes, class_weights))
print("Class weights (1-5 stars):", {k+1: round(v, 2) for k, v in class_weight_dict.items()})

# Baseline: Logistic Regression
print("BASELINE: Logistic Regression (with class weights)")

lr = LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42,
    n_jobs=-1
)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)

print(f"Test Accuracy: {accuracy_score(y_test, y_pred_lr):.4f}")
print("\nClassification Report (Logistic Regression):\n")
print(classification_report(y_test, y_pred_lr, target_names=['1','2','3','4','5']))

# Neural Network (regularized + class weights)
print("NEURAL NETWORK (MLP) – class weights + L2 + early stopping")

X_train_dense = X_train.toarray()
X_test_dense  = X_test.toarray()

model = models.Sequential([
    layers.Dense(128, activation='relu',
                 kernel_regularizer=regularizers.l2(1e-4),
                 input_shape=(X_train_dense.shape[1],)),
    layers.Dropout(0.4),
    layers.Dense(64, activation='relu',
                 kernel_regularizer=regularizers.l2(1e-4)),
    layers.Dropout(0.4),
    layers.Dense(32, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(5, activation='softmax')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

early_stop = callbacks.EarlyStopping(
    monitor='val_loss', patience=3, restore_best_weights=True, verbose=1
)
reduce_lr = callbacks.ReduceLROnPlateau(
    monitor='val_loss', factor=0.5, patience=2, min_lr=1e-5, verbose=1
)

history = model.fit(
    X_train_dense, y_train,
    epochs=25,
    batch_size=32,
    validation_split=0.2,
    class_weight=class_weight_dict,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

test_loss, test_acc = model.evaluate(X_test_dense, y_test, verbose=0)
print(f"\nNeural Net Test Accuracy: {test_acc:.4f}")

y_pred_nn = np.argmax(model.predict(X_test_dense, verbose=0), axis=1)
print("\nClassification Report (Neural Network):\n")
print(classification_report(y_test, y_pred_nn, target_names=['1','2','3','4','5']))

# Side-by-side confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

cm_lr = confusion_matrix(y_test, y_pred_lr)
sns.heatmap(cm_lr, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=[1,2,3,4,5], yticklabels=[1,2,3,4,5])
axes[0].set_title('Logistic Regression\n(with class weights)')
axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('True')

cm_nn = confusion_matrix(y_test, y_pred_nn)
sns.heatmap(cm_nn, annot=True, fmt='d', cmap='Greens', ax=axes[1],
            xticklabels=[1,2,3,4,5], yticklabels=[1,2,3,4,5])
axes[1].set_title('Neural Network (MLP)\n(with class weights + regularization)')
axes[1].set_xlabel('Predicted')
axes[1].set_ylabel('True')

plt.tight_layout()
plt.show()

# Learning curves
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Loss Curves')
plt.xlabel('Epoch')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Accuracy Curves')
plt.xlabel('Epoch')
plt.legend()
plt.tight_layout()
plt.show()
```

### Model Evaluation: Addressing Imbalance & Comparing Approaches
We improved the modeling pipeline in three concrete ways:

1. **Baseline comparison** – A class-weighted Logistic Regression provides a strong, interpretable reference point.
2. **Actual imbalance handling** – Balanced class weights are passed to both the logistic regression and the neural network so minority classes (1–3 star) are no longer ignored.
3. **Stronger regularization** – L2 weight decay, higher dropout rates (0.3–0.4), early stopping, and learning-rate reduction on plateau significantly reduce overfitting.

**Key observations after the upgrades:**
* Class weights force the models to pay attention to the scarce 1-, 2-, and 3-star reviews. Both models now produce non-zero precision/recall on the minority classes (previously 0.00).
* The Logistic Regression baseline is competitive and often more stable than the neural net on this small, highly imbalanced dataset.
* Learning curves show that early stopping successfully prevents the neural net from continuing to overfit after the validation loss stops improving.
* Residual difficulty remains on the most nuanced ratings (2- and 3-star). Further gains would likely come from richer text representations (e.g., sentence transformers) or targeted data collection for the minority classes rather than from deeper architectures alone.

**Business takeaway:** Even a simple, well-calibrated linear model with class weighting already delivers usable automated rating prediction and surfaces the same product-feature signals discovered in the NLP section. The neural network can be used when additional capacity is needed, but only after imbalance is explicitly addressed.
